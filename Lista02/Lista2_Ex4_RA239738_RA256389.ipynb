```python
# Dicionário que contem as conexões entre as cidades da Romênia,distância entre uma cidade e as possíveis cidades vizinhas,e a distância heurística (linha reta) entre a cidade e o objetivo (Bucharest).
mapa_Romenia ={
    'Arad': {'prox_cidade':{'Zerind': 75, 'Sibiu': 140, 'Timisoara': 118}, 'h': 366},
    'Bucharest': {'prox_cidade':{'Urziceni': 85, 'Giurgiu': 90, 'Pitesti': 101, 'Fagaras': 211}, 'h': 0},
    'Craiova': {'prox_cidade':{'Drobeta': 120, 'Rimnicu Vilcea': 146, 'Pitesti': 138}, 'h': 160},
    'Drobeta': {'prox_cidade':{'Mehadia': 75, 'Craiova': 120}, 'h': 242},
    'Eforie': {'prox_cidade':{'Hirsova': 86}, 'h': 161},
    'Fagaras': {'prox_cidade':{'Sibiu': 99, 'Bucharest': 211}, 'h': 176},
    'Giurgiu': {'prox_cidade':{'Bucharest': 90}, 'h': 77},
    'Hirsova': {'prox_cidade':{'Urziceni': 98, 'Eforie': 86}, 'h': 151},
    'Iasi': {'prox_cidade':{'Neamt': 87, 'Vaslui': 92}, 'h': 226},
    'Lugoj': {'prox_cidade':{'Timisoara': 111, 'Mehadia': 70}, 'h': 244},
    'Mehadia': {'prox_cidade':{'Lugoj': 70, 'Drobeta': 75}, 'h': 241},
    'Neamt': {'prox_cidade':{'Iasi': 87}, 'h': 234},
    'Oradea': {'prox_cidade':{'Zerind': 71, 'Sibiu': 151}, 'h': 380},
    'Pitesti': {'prox_cidade':{'Rimnicu Vilcea': 97, 'Craiova': 138, 'Bucharest': 101}, 'h': 100},
    'Rimnicu Vilcea': {'prox_cidade':{'Sibiu': 80, 'Pitesti': 97, 'Craiova': 146}, 'h': 193},
    'Sibiu': {'prox_cidade':{'Arad': 140, 'Oradea': 151, 'Fagaras': 99, 'Rimnicu Vilcea': 80}, 'h': 253},
    'Timisoara': {'prox_cidade':{'Arad': 118, 'Lugoj': 111}, 'h': 329},
    'Urziceni': {'prox_cidade':{'Bucharest': 85, 'Hirsova': 98, 'Vaslui': 142}, 'h': 80},
    'Vaslui': {'prox_cidade':{'Iasi': 92, 'Urziceni': 142}, 'h': 199},
    'Zerind': {'prox_cidade':{'Arad': 75, 'Oradea': 71}, 'h': 374}  

}
```

```python
# Função que retorna a próxima cidade a ser visitada com base no algoritmo A*
def proxima_cidade(cidade_atual, nos_visitados=[]):
    # Opções de cidades a serem visitadas a partir da cidade atual
    opcoes = mapa_Romenia[cidade_atual]['prox_cidade']

    # Exclui vizinhos que já foram visitados
    candidatos = {cidade: dist for cidade, dist in opcoes.items() if cidade not in nos_visitados}

    # Calcula f para cada cidade
    f = {cidade: dist + mapa_Romenia[cidade]['h']
              for cidade, dist in candidatos.items()}
    
    # Retorna a cidade com o menor valor de f,g e h
    prox_cidade = min(f, key=f.get)
    f = f[prox_cidade]
    g = mapa_Romenia[cidade_atual]['prox_cidade'][prox_cidade]
    h = mapa_Romenia[prox_cidade]['h']
    return prox_cidade,f,g,h
```

```python
proxima_cidade('Arad')  # Exemplo de uso da função
```

```python
# Função que implementa o Algoritmo A*
def Astar(cidade_atual,nos_percorridos=None):
    # Inicializa a lista de nós percorridos se não for fornecida
    if nos_percorridos is None:
        nos_percorridos = []

    # Adiciona a cidade atual aos nós percorridos
    nos_percorridos.append(cidade_atual)
    # Obtém a próxima cidade, f, g e h
    next_city, f, g, h = proxima_cidade(cidade_atual,nos_percorridos)
    # Adiciona a cidade atual aos nós percorridos
    nos_percorridos.append(cidade_atual)
    # Printa a ação tomada
    print(f"De {cidade_atual} vá para {next_city} com f={f}, g={g}, h={h}")
    # Se atingiu o a cidade objetivo sai da função, se não, chama a função novamente com a próxima cidade
    if next_city == 'Bucharest':
        return
    else:
        Astar(next_city,nos_percorridos)
```
