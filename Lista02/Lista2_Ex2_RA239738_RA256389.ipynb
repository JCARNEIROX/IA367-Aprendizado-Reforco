```python
from collections import deque
```

```python
# Função para verificar se um estado é válido
def valido(m1, c1, m2, c2):
    # valores não podem ser negativos
    if m1 < 0 or c1 < 0 or m2 < 0 or c2 < 0:
        return False
    # restrição: missionários não podem ser menos que canibais em uma margem (se houver missionários)
    if m1 > 0 and m1 < c1:
        return False
    if m2 > 0 and m2 < c2:
        return False
    return True

# Função que gera sucessores de um estado
def sucessores(estado, cap_min=1, cap_max=2):
    """
    Gera próximos estados válidos.
    Barco leva entre cap_min e cap_max pessoas (padrão 1 ou 2).
    """
    m1, c1, m2, c2, lado = estado
    result = []
    # todos os pares (dm,dc) possíveis
    moves = []
    for dm in range(0, cap_max + 1):
        for dc in range(0, cap_max + 1):
            if dm + dc == 0:
                continue
            if cap_min <= dm + dc <= cap_max:
                moves.append((dm, dc))
    # gerar novos estados
    for dm, dc in moves:
        if lado == 'esquerda':
            novo = (m1 - dm, c1 - dc, m2 + dm, c2 + dc, 'direita')
        else:
            novo = (m1 + dm, c1 + dc, m2 - dm, c2 - dc, 'esquerda')
        if valido(*novo[:4]):
            result.append(novo)
    return result

# Algoritmo BFS para encontrar solução
def bfs(m, c, cap_min=1, cap_max=2):
    ini = (m, c, 0, 0, 'esquerda')
    goal = (0, 0, m, c, 'direita')
    q = deque([(ini, [ini])])
    vis = set([ini])

    while q:
        s, path = q.popleft()
        if s == goal:
            return path
        for nxt in sucessores(s, cap_min, cap_max):
            if nxt not in vis:
                vis.add(nxt)
                q.append((nxt, path + [nxt]))
    return None

def imprimir_caminho(path):
    for i in range(len(path)):
        m1, c1, m2, c2, lado = path[i]
        print(f"Passo {i}:  Esq(M={m1}, C={c1})  Dir(M={m2}, C={c2})  Barco: {lado}")
        if i+1 < len(path):
            m1n, c1n, m2n, c2n, ladon = path[i+1]
            # quem se moveu
            dm = abs(m1 - m1n)
            dc = abs(c1 - c1n)
            sentido = "→ direita" if lado == 'esquerda' else "→ esquerda"
            print(f"         Move: {dm}M e {dc}C {sentido}")
```

```python
# ---------- Executar ----------
sol = bfs(3, 3, cap_min=1, cap_max=2)
if sol:
    print("Solução encontrada!\n")
    imprimir_caminho(sol)
else:
    print("Nenhuma solução encontrada.")
```
